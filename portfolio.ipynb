```python
import json
import os
import zipfile

CONFIG = {
    "name": "Pratyay Kumar Biswas",
    "role": "Data Scientist",
    "tagline": "I turn messy data into decisions — models, pipelines, and the occasional dashboard people actually use.",
    "email": "pratyaybiswas66@gmail.com",
    "location": "Remote / Open to relocation",
    "socials": [
        {"label": "GitHub", "url": "https://github.com/Pratyay-B1swas"},
        {"label": "LinkedIn", "url": "https://www.linkedin.com/in/pratyay-kumar-biswas-807997285/"},
    ],

    "about": ("Motivated Computer Science & Engineering undergraduate with a strong foundation in Python, C/C++, and SQL, and hands-on exposure to cybersecurity fundamentals through government training. Eager to apply strong problem-solving skills and a fast learning curve to an entry-level role in software development"
    ),
    "tools": [
        "Python", "SQL", "Pandas", "NumPy", "Scikit-Learn", "TensorFlow", "PyTorch",
         "C", "C++", "PowerBI"
    ],
    "skills": [
        {"name": "Python", "level": 90},
        {"name": "SQL", "level": 90},
        {"name": "Machine Learning", "level": 50},
        {"name": "Statistics", "level": 80},
        {"name": "Data Engineering", "level": 75},
        {"name": "Visualization", "level": 90},
    ],

    "stats": [
        {"value": "10+", "label": "Completed peojects"},
        {"value": "1", "label": "Thesis Paper Complete"},
    ],

    "projects": [
        {
            "title": "saas-revenue-churn-intelligence",
            "brief": "An end-to-end Data Analytics & Machine Learning project built in Google Colab to predict customer churn risk, quantify potential Monthly Recurring Revenue (MRR) loss, and generate automated retention strategies.",
            "stack": ["Python", "Pandas" , "NumPy", "Scikit-Learn", "Matplotlib", "Seaborn"],
            "metric_label": "Calculates granular 0–100% risk scores for every active account ",
            "link": "https://github.com/Pratyay-B1swas/saas-revenue-churn-intelligence",
        },
        {
            "title": "Hyper-local-donation-website",
            "brief": "Donation Center.",
            "stack": ["Vite ", "TypeScript", "React", "shadcn-ui", "Tailwind", "CSS"],
            "link": "https://github.com/Pratyay-B1swas/Hyper-local-donation-website/tree/main",
        },
        {
            "title": "E-commerce-Customer-Segmentation",
            "brief": "This project focuses on performing Customer Segmentation for an e-commerce retailer to identify distinct customer "
                     "groups. By understanding customer behavior based on their transaction history, targeted marketing strategies can be"
                     "developed to maximize customer lifetime value (CLV) and optimize marketing budget allocation.",
            "stack": ["Python", "RFM Analysis", "K-Means Clustering"],
            "link": "https://github.com/Pratyay-B1swas/E-commerce-Customer-Segmentation",
        },
        {
            "title": "SocialSentimentTracker",
            "brief": "Analyzing people's opinions on a specific social event.",
            "stack": ["Python", "News API", "Pandas", "TextBlob", "Matplotlib/Seaborn"],
            "link": "https://github.com/Pratyay-B1swas/SocialSentimentTracker-",
        },
    ],

    "experience": [
        {
            "role": "Cyber Security KPI Training", "org": "Government Program ", "period": "2026", "points": []
        }
    ],

    "education": [
        {"title": "S.S.C", "org": "Keshabpur Pilot School & College", "year": "2019"},
        {"title": "H.S.C", "org": "Hamidpur Al-Hera College", "year": "2021"},
        {"title": "B.S.C", "org": "Northern University Bangladesh", "year": "2023 - current"},
    ],
    "certifications": [
        "Data Science & Machine Learning with Python for Beginners — Ostad",
    ],
}

# =========================================================================
# 2. TEMPLATE CODE — you generally don't need to edit below this line
# =========================================================================

CSS = """
:root{
  --ink:#12181F;
  --ink-soft:#1A222C;
  --paper:#F7F5F0;
  --paper-dim:#EAE6DC;
  --line:#2A323D;
  --line-soft:rgba(247,245,240,0.12);
  --teal:#4FD1C5;
  --amber:#E8A33D;
  --slate:#8B96A5;
  --text:#E7E9EC;
  --font-display: 'Fraunces', Georgia, serif;
  --font-body: 'IBM Plex Sans', -apple-system, sans-serif;
  --font-mono: 'IBM Plex Mono', ui-monospace, monospace;
}
*{
box-sizing:border-box;
}
html{
scroll-behavior:smooth;
}
body{
  margin:0;
  background:
    radial-gradient(circle at 1.5px 1.5px, var(--line-soft) 1.5px, transparent 0) 0 0/28px 28px,
    var(--ink);
  color:var(--text);
  font-family:var(--font-body);
  line-height:1.6;
  -webkit-font-smoothing:antialiased;
}
a{
color:var(--teal);
text-decoration:none;
}
a:hover{
text-decoration:underline;
}
a:focus-visible, button:focus-visible{
outline:2px solid var(--teal);
outline-offset:3px;
}

.wrap{
max-width:1100px;
margin:0 auto;
padding:0 32px;
}

/* ---------- NAV ---------- */
nav{
  position:sticky;
top:0;
z-index:50;
  background:rgba(18,24,31,0.9);
backdrop-filter:blur(8px);
  border-bottom:1px solid var(--line);
}
nav .wrap{
display:flex;
align-items:center;
justify-content:space-between;
height:64px;
}
.brand{
font-family:var(--font-mono);
font-size:15px;
color:var(--text);
}
.brand span{
color:var(--teal);
}
nav ul{
list-style:none;
display:flex;
gap:28px;
margin:0;
padding:0;
}
nav a{
color:var(--slate);
font-size:15px;
}
nav a:hover{
color:var(--text);
text-decoration:none;
}

/* ---------- HERO ---------- */
.hero{
padding:88px 0 56px;
}
.hero-grid{
display:grid;
grid-template-columns:1.1fr 0.9fr;
gap:56px;
align-items:center;
}

.avatar-ring{
  position:relative;
width:88px;
height:88px;
margin-bottom:24px;
}
.avatar-ring svg{
position:absolute;
inset:0;
transform:rotate(-90deg);
}
.avatar-ring .ring-bg{
fill:none;
stroke:var(--line);
stroke-width:3;
}
.avatar-ring .ring-fill{
  fill:none;
stroke:var(--teal);
stroke-width:3;
stroke-linecap:round;
  stroke-dasharray:251;
stroke-dashoffset:251;
  transition:stroke-dashoffset 1.4s ease;
}
.avatar-ring .monogram{
  position:absolute;
inset:8px;
border-radius:50%;
  background:var(--ink-soft);
display:flex;
align-items:center;
justify-content:center;
  font-family:var(--font-display);
font-size:26px;
font-weight:600;
color:var(--paper);
}

.hero-eyebrow{
  font-family:var(--font-mono);
font-size:13px;
color:var(--teal);
margin:0 0 18px;
}
.hero h1{
  font-family:var(--font-display);
  font-size:56px;
line-height:1.05;
margin:0 0 20px;
font-weight:600;
  letter-spacing:-0.01em;
}
.hero p.tagline{
color:var(--slate);
font-size:18px;
max-width:46ch;
margin:0 0 32px;
}
.hero-actions{
display:flex;
gap:14px;
flex-wrap:wrap;
}
.btn{
  display:inline-block;
padding:12px 22px;
border-radius:3px;
font-size:15px;
  border:1px solid var(--line);
font-family:var(--font-body);
}
.btn-primary{
background:var(--teal);
color:var(--ink);
border-color:var(--teal);
font-weight:600;
}
.btn-primary:hover{
background:#3fc0b4;
text-decoration:none;
}
.btn-ghost{
color:var(--text);
}
.btn-ghost:hover{
border-color:var(--teal);
color:var(--teal);
text-decoration:none;
}

.hero-chart-box{
  background:var(--ink-soft);
border:1px solid var(--line);
border-radius:4px;
padding:18px;
}
.hero-chart-box .caption{
font-family:var(--font-mono);
font-size:12px;
color:var(--slate);
margin:10px 2px 0;
}

/* ---------- STATS BAR ---------- */
.stats-bar{
border-top:1px solid var(--line);
border-bottom:1px solid var(--line);
}
.stats-grid{
  display:grid;
grid-template-columns:repeat(4, 1fr);
  padding:36px 0;
}
.stat{
text-align:left;
padding:0 24px;
border-left:1px solid var(--line);
}
.stat:first-child{
border-left:none;
padding-left:0;
}
.stat .value{
  font-family:var(--font-display);
font-size:38px;
font-weight:600;
color:var(--teal);
  display:block;
}
.stat .label{
font-size:13px;
color:var(--slate);
margin-top:4px;
}

/* ---------- EDUCATION / CERTS ---------- */
.edu-cert-grid{
display:grid;
grid-template-columns:1fr 1fr;
gap:48px;
}
.edu-item{
padding:14px 0;
border-top:1px solid var(--line);
}
.edu-item:first-child{
border-top:none;
}
.edu-item h3{
font-family:var(--font-display);
font-size:18px;
margin:0 0 2px;
font-weight:600;
}
.edu-item p{
color:var(--slate);
font-size:14px;
margin:0;
}
.edu-item .year{
font-family:var(--font-mono);
color:var(--amber);
font-size:12px;
}
.cert-badges{
display:flex;
flex-direction:column;
gap:10px;
}
.cert-badge{
  display:flex;
align-items:center;
gap:10px;
  border:1px solid var(--line);
border-radius:3px;
padding:10px 14px;
  font-size:14px;
color:var(--text);
background:var(--ink-soft);
}
.cert-badge::before{
  content:"\2713";
color:var(--teal);
font-family:var(--font-mono);
font-weight:600;
}

/* ---------- SECTION SCAFFOLD (notebook cell) ---------- */
section{
padding:64px 0;
}
.cell{
  border-top:1px solid var(--line);
  padding-top:40px;
}
.cell-marker{
  font-family:var(--font-mono);
font-size:12px;
color:var(--slate);
margin-bottom:14px;
}
.cell-marker b{
color:var(--amber);
}
h2{
  font-family:var(--font-display);
font-size:32px;
font-weight:600;
margin:0 0 24px;
}

/* ---------- ABOUT ---------- */
.about-grid{
display:grid;
grid-template-columns:1.3fr 1fr;
gap:48px;
}
.about-grid p{
color:var(--text);
font-size:17px;
max-width:60ch;
}
.tools-list{
display:flex;
flex-wrap:wrap;
gap:8px;
margin-top:8px;
}
.tools-list span{
  font-family:var(--font-mono);
font-size:13px;
color:var(--paper);
  background:var(--ink-soft);
border:1px solid var(--line);
  padding:5px 10px;
border-radius:3px;
}

.skills-grid{
display:grid;
grid-template-columns:1fr 1fr;
gap:40px;
align-items:center;
}
.skill-bars{
display:flex;
flex-direction:column;
gap:16px;
}
.skill-row{
display:grid;
grid-template-columns:140px 1fr 40px;
align-items:center;
gap:12px;
}
.skill-row .name{
font-size:14px;
color:var(--text);
}
.skill-row .track{
height:8px;
background:var(--ink-soft);
border:1px solid var(--line);
border-radius:4px;
overflow:hidden;
}
.skill-row .fill{
height:100%;
background:var(--teal);
width:0%;
transition:width 1.1s ease;
}
.skill-row .val{
font-family:var(--font-mono);
font-size:12px;
color:var(--slate);
text-align:right;
}

.project{
  display:grid;
grid-template-columns:1fr 1fr;
gap:40px;
  padding:36px 0;
border-top:1px solid var(--line);
}
.project:first-of-type{
border-top:none;
padding-top:0;
}
.project-body h3{
font-family:var(--font-display);
font-size:24px;
margin:0 0 12px;
font-weight:600;
}
.project-body p{
color:var(--slate);
font-size:15px;
max-width:52ch;
}
.project-stack{
display:flex;
flex-wrap:wrap;
gap:6px;
margin:16px 0;
}
.project-stack span{
  font-family:var(--font-mono);
font-size:12px;
color:var(--slate);
  border:1px solid var(--line);
padding:3px 8px;
border-radius:3px;
}
.metric{
margin:18px 0;
}
.metric .value{
font-family:var(--font-display);
font-size:34px;
color:var(--amber);
font-weight:600;
}
.metric .label{
font-size:13px;
color:var(--slate);
}
.project-link{
font-family:var(--font-mono);
font-size:14px;
}
.project-chart{
background:var(--ink-soft);
border:1px solid var(--line);
border-radius:4px;
padding:14px;
}
.project-chart .caption{
font-family:var(--font-mono);
font-size:12px;
color:var(--slate);
margin-top:8px;
}

/* ---------- EXPERIENCE (this one IS a real sequence, numbering is earned) ---------- */
.timeline{
display:flex;
flex-direction:column;
gap:0;
}
.tl-item{
display:grid;
grid-template-columns:32px 1fr;
gap:20px;
padding:24px 0;
border-top:1px solid var(--line);
}
.tl-item:first-child{
border-top:none;
}
.tl-num{
font-family:var(--font-mono);
color:var(--teal);
font-size:14px;
padding-top:2px;
}
.tl-item h3{
font-family:var(--font-display);
font-size:20px;
margin:0 0 2px;
font-weight:600;
}
.tl-org{
color:var(--slate);
font-size:14px;
margin:0 0 12px;
}
.tl-org .period{
color:var(--amber);
font-family:var(--font-mono);
font-size:12px;
margin-left:8px;
}
.tl-item ul{
margin:0;
padding-left:18px;
color:var(--text);
font-size:15px;
}
.tl-item li{
margin-bottom:6px;
}

/* ---------- CONTACT ---------- */
.contact-box{
  border:1px solid var(--line);
border-radius:4px;
padding:40px;
  background:var(--ink-soft);
display:flex;
justify-content:space-between;
align-items:center;
flex-wrap:wrap;
gap:24px;
}
.contact-box h2{
margin:0 0 6px;
}
.contact-box p{
color:var(--slate);
margin:0;
}
.social-links{
display:flex;
gap:18px;
}
.social-links a{
font-family:var(--font-mono);
font-size:14px;
}

footer{
padding:32px 0 48px;
border-top:1px solid var(--line);
}
footer .wrap{
display:flex;
justify-content:space-between;
color:var(--slate);
font-size:13px;
font-family:var(--font-mono);
}

/* ---------- RESPONSIVE ---------- */
@media (max-width: 860px){
  .hero-grid, .about-grid, .skills-grid, .project, .edu-cert-grid{
grid-template-columns:1fr;
}
  .stats-grid{
grid-template-columns:1fr 1fr;
row-gap:24px;
}
  .stat{
border-left:none;
padding-left:0;
}
  .stat:nth-child(even){
padding-left:24px;
border-left:1px solid var(--line);
}
  .hero h1{
font-size:38px;
}
  .wrap{
padding:0 20px;
}
  .contact-box{
flex-direction:column;
align-items:flex-start;
}
}

@media (prefers-reduced-motion: reduce){
  html{
scroll-behavior:auto;
}
  .skill-row .fill{
transition:none;
}
}
"""

JS = """
document.addEventListener('DOMContentLoaded', function () {
  var data = JSON.parse(document.getElementById('site-data').textContent);

  // ---- animate skill bars once, on load ----
  document.querySelectorAll('.skill-row .fill').forEach(function (el) {
    var target = el.getAttribute('data-level');
    requestAnimationFrame(function () {
      setTimeout(function () { el.style.width = target + '%'; }, 150);
    });
  });

  // ---- animate avatar ring (fills to a fixed "experience" proportion) ----
  var ringFill = document.querySelector('.ring-fill');
  if (ringFill) {
    var pct = parseFloat(ringFill.getAttribute('data-pct')) || 80;
    var offset = 251 - (251 * pct / 100);
    requestAnimationFrame(function () {
      setTimeout(function () { ringFill.style.strokeDashoffset = offset; }, 200);
    });
  }

  // ---- animate stat numbers counting up ----
  document.querySelectorAll('.stat .value').forEach(function (el) {
    var raw = el.textContent.trim();
    var match = raw.match(/^(-?\\d+(?:\\.\\d+)?)(.*)$/);
    if (!match) return;
    var target = parseFloat(match[1]);
    var suffix = match[2];
    var decimals = match[1].includes('.') ? match[1].split('.')[1].length : 0;
    var current = 0;
    var steps = 30;
    var increment = target / steps;
    el.textContent = '0' + suffix;
    var count = 0;
    var timer = setInterval(function () {
      count += 1;
      current += increment;
      if (count >= steps) {
        el.textContent = target.toFixed(decimals) + suffix;
        clearInterval(timer);
      } else {
        el.textContent = current.toFixed(decimals) + suffix;
      }
    }, 35);
  });

  var plotLayout = {
    paper_bgcolor: 'rgba(0,0,0,0)',
    plot_bgcolor: 'rgba(0,0,0,0)',
    font: { color: '#8B96A5', family: 'IBM Plex Mono', size: 11 },
    margin: { l: 40, r: 20, t: 10, b: 30 },
    xaxis: { gridcolor: '#2A323D', zeroline: false },
    yaxis: { gridcolor: '#2A323D', zeroline: false },
  };
  var plotConfig = { displayModeBar: false, responsive: true };

  // ---- hero radar-ish line chart (skills as a signal) ----
  var skillNames = data.skills.map(function (s) { return s.name; });
  var skillLevels = data.skills.map(function (s) { return s.level; });
  Plotly.newPlot('hero-chart', [{
    x: skillNames, y: skillLevels, type: 'scatter', mode: 'lines+markers',
    line: { color: '#4FD1C5', width: 2, shape: 'spline' },
    marker: { color: '#E8A33D', size: 7 },
  }], Object.assign({}, plotLayout, { margin: { l: 30, r: 10, t: 10, b: 40 } }), plotConfig);

  // ---- skills radar chart in Skills section ----
  Plotly.newPlot('skills-radar', [{
    type: 'scatterpolar',
    r: skillLevels.concat([skillLevels[0]]),
    theta: skillNames.concat([skillNames[0]]),
    fill: 'toself',
    fillcolor: 'rgba(79,209,197,0.15)',
    line: { color: '#4FD1C5' },
  }], {
    paper_bgcolor: 'rgba(0,0,0,0)',
    font: { color: '#8B96A5', family: 'IBM Plex Mono', size: 11 },
    polar: {
      bgcolor: 'rgba(0,0,0,0)',
      radialaxis: { visible: true, range: [0, 100], gridcolor: '#2A323D', color: '#8B96A5' },
      angularaxis: { gridcolor: '#2A323D', color: '#E7E9EC' },
    },
    margin: { l: 40, r: 40, t: 30, b: 30 },
    showlegend: false,
  }, plotConfig);

  // ---- per-project charts ----
  data.projects.forEach(function (p, i) {
    var elId = 'project-chart-' + i;

    if (p.chart_type === 'heatmap') {
      Plotly.newPlot(elId, [{
        z: p.chart_z, x: p.chart_x, y: p.chart_y,
        type: 'heatmap', colorscale: [[0, '#1A222C'], [1, '#4FD1C5']],
        showscale: false, texttemplate: '%{z}', textfont: { color: '#12181F', size: 13 },
      }], Object.assign({}, plotLayout, {
        margin: { l: 90, r: 10, t: 10, b: 60 },
        xaxis: { side: 'bottom', gridcolor: '#2A323D' },
      }), plotConfig);
      return;
    }

    if (p.chart_type === 'roc') {
      Plotly.newPlot(elId, [
        {
          x: p.chart_x, y: p.chart_y, type: 'scatter', mode: 'lines',
          fill: 'tozeroy', fillcolor: 'rgba(232,163,61,0.15)',
          line: { color: '#E8A33D', width: 2 }, name: 'model',
        },
        {
          x: [0, 1], y: [0, 1], type: 'scatter', mode: 'lines',
          line: { color: '#2A323D', width: 1, dash: 'dot' }, name: 'chance',
        },
      ], Object.assign({}, plotLayout, { showlegend: false }), plotConfig);
      return;
    }

    var trace = {
      x: p.chart_x, y: p.chart_y,
      type: p.chart_type === 'line' ? 'scatter' : 'bar',
      mode: p.chart_type === 'line' ? 'lines+markers' : undefined,
      marker: { color: p.chart_type === 'line' ? '#E8A33D' : '#4FD1C5' },
      line: p.chart_type === 'line' ? { color: '#E8A33D', width: 2 } : undefined,
    };
    Plotly.newPlot(elId, [trace], plotLayout, plotConfig);
  });
});
"""

def render_skill_rows(skills):
    rows = []
    for s in skills:
        rows.append(f"""
        <div class="skill-row">
          <div class="name">{s['name']}</div>
          <div class="track"><div class="fill" data-level="{s['level']}"></div></div>
          <div class="val">{s['level']}%</div>
        </div>"""
)
    return "".join(rows)


def render_tools(tools):
    return "".join(f"<span>{t}</span>" for t in tools)


def render_projects(projects):
    blocks = []
    for i, p in enumerate(projects):
        stack = "".join(f"<span>{s}</span>" for s in p["stack"])

        metric_html = ""
        if "metric_value" in p and "metric_label" in p:
            metric_html = f"""
            <div class="metric">
              <div class="value">{p['metric_value']}</div>
              <div class="label">{p['metric_label']}</div>
            </div>
            """

        chart_html = ""
        if "chart_type" in p and "chart_label" in p: # Assuming chart_type implies other chart data
             chart_html = f"""
            <div class="project-chart">
              <div id="project-chart-{i}" style="height:220px;"></div>
              <div class="caption">{p['chart_label']}</div>
            </div>
            """

        blocks.append(f"""
        <div class="project">
          <div class="project-body">
            <h3>{p['title']}</h3>
            <p>{p['brief']}</p>
            <div class="project-stack">{stack}</div>
            {metric_html}
            <a class="project-link" href="{p['link']}" target="_blank" rel="noopener">View repository &#8594;</a>
          </div>
          {chart_html}
        </div>"""
)
    return "".join(blocks)


def render_experience(experience):
    items = []
    for i, e in enumerate(experience, start=1):
        points = "".join(f"<li>{pt}</li>" for pt in e["points"])
        items.append(f"""
        <div class="tl-item">
          <div class="tl-num">{i:02d}</div>
          <div>
            <h3>{e['role']}</h3>
            <p class="tl-org">{e['org']}<span class="period">{e['period']}</span></p>
            <ul>{points}</ul>
          </div>
        </div>"""
)
    return "".join(items)


def render_socials(socials):
    return "".join(f'<a href="{s["url"]}" target="_blank" rel="noopener">{s["label"]}</a>' for s in socials)


def render_stats(stats):
    blocks = []
    for s in stats:
        blocks.append(f"""
        <div class="stat">
          <span class="value">{s['value']}</span>
          <span class="label">{s['label']}</span>
        </div>"""
)
    return "".join(blocks)


def render_education(education):
    items = []
    for e in education:
        items.append(f"""
        <div class="edu-item">
          <h3>{e['title']} <span class="year">{e['year']}</span></h3>
          <p>{e['org']}</p>
        </div>"""
)
    return "".join(items)


def render_certifications(certs):
    return "".join(f'<div class="cert-badge">{c}</div>' for c in certs)


def build_html(cfg):
    site_data = json.dumps({
        "skills": cfg["skills"],
        "projects": cfg["projects"],
    })
    initials = "".join(w[0].upper() for w in cfg["name"].split()[:2])

    html = f"""<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>{cfg['name']} — {cfg['role']}</title>
<meta name="description" content="{cfg['tagline']}">
<link rel="preconnect" href="https://fonts.googleapis.com">
<link href="https://fonts.googleapis.com/css2?family=Fraunces:wght@500;600;700&family=IBM+Plex+Sans:wght@400;500;600&family=IBM+Plex+Mono:wght@400;500&display=swap" rel="stylesheet">
<script src="https://cdn.plot.ly/plotly-2.32.0.min.js"></script>
<style>{CSS}</style>
</head>
<body>

<nav>
  <div class="wrap">
    <div class="brand">{cfg['name'].split()[0].lower()}<span>.py</span></div>
    <ul>
      <li><a href="#about">about</a></li>
      <li><a href="#skills">skills</a></li>
      <li><a href="#projects">projects</a></li>
      <li><a href="#experience">experience</a></li>
      <li><a href="#education">education</a></li>
      <li><a href="#contact">contact</a></li>
    </ul>
  </div>
</nav>

<header class="hero">
  <div class="wrap hero-grid">
    <div>
      <div class="avatar-ring">
        <svg viewBox="0 0 88 88">
          <circle class="ring-bg" cx="44" cy="44" r="40"></circle>
          <circle class="ring-fill" data-pct="82" cx="44" cy="44" r="40"></circle>
        </svg>
        <div class="monogram">{initials}</div>
      </div>
      <p class="hero-eyebrow">{cfg['location']}</p>
      <h1>{cfg['name']}<br>{cfg['role']}</h1>
      <p class="tagline">{cfg['tagline']}</p>
      <div class="hero-actions">
        <a class="btn btn-primary" href="#projects">See the work</a>
        <a class="btn btn-ghost" href="#contact">Get in touch</a>
      </div>
    </div>
    <div class="hero-chart-box">
      <div id="hero-chart" style="height:260px;"></div>
      <div class="caption">fig. 1 — core skill signal, self-rated</div>
    </div>
  </div>
</header>

<div class="stats-bar">
  <div class="wrap stats-grid">{render_stats(cfg['stats'])}</div>
</div>

<section id="about">
  <div class="wrap cell">
    <div class="cell-marker">In [<b>1</b>]: <span>about()</span></div>
    <div class="about-grid">
      <div>
        <h2>About</h2>
        <p>{cfg['about']}</p>
      </div>
      <div>
        <p class="cell-marker">Out[<b>1</b>]: tools I reach for</p>
        <div class="tools-list">{render_tools(cfg['tools'])}</div>
      </div>
    </div>
  </div>
</section>

<section id="skills">
  <div class="wrap cell">
    <div class="cell-marker">In [<b>2</b>]: <span>plot_skills()</span></div>
    <h2>Skills</h2>
    <div class="skills-grid">
      <div class="skill-bars">{render_skill_rows(cfg['skills'])}</div>
      <div id="skills-radar" style="height:320px;"></div>
    </div>
  </div>
</section>

<section id="projects">
  <div class="wrap cell">
    <div class="cell-marker">In [<b>3</b>]: <span>load_projects()</span></div>
    <h2>Projects</h2>
    {render_projects(cfg['projects'])}
  </div>
</section>

<section id="experience">
  <div class="wrap cell">
    <div class="cell-marker">In [<b>4</b>]: <span>timeline()</span></div>
    <h2>Experience</h2>
    <div class="timeline">{render_experience(cfg['experience'])}</div>
  </div>
</section>

<section id="education">
  <div class="wrap cell">
    <div class="cell-marker">In [<b>5</b>]: <span>credentials()</span></div>
    <h2>Education &amp; Certifications</h2>
    <div class="edu-cert-grid">
      <div>{render_education(cfg['education'])}</div>
      <div class="cert-badges">{render_certifications(cfg['certifications'])}</div>
    </div>
  </div>
</section>

<section id="contact">
  <div class="wrap cell">
    <div class="cell-marker">In [<b>6</b>]: <span>contact()</span></div>
    <div class="contact-box">
      <div>
        <h2>Let's talk</h2>
        <p>{cfg['email']}</p>
      </div>
      <div class="social-links">{render_socials(cfg['socials'])}</div>
    </div>
  </div>
</section>

<footer>
  <div class="wrap">
    <span>&copy; {cfg['name']}</span>
    <span>built with python</span>
  </div>
</footer>

<script id="site-data" type="application/json">{site_data}</script>
<script>{JS}</script>
</body>
</html>
"""
    return html


def main():
    out_dir = "portfolio_site"
    os.makedirs(out_dir, exist_ok=True)

    html = build_html(CONFIG)
    index_path = os.path.join(out_dir, "index.html")
    with open(index_path, "w", encoding="utf-8") as f:
        f.write(html)

    zip_path = "portfolio_site.zip"
    with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
        zf.write(index_path, arcname="index.html")

    print(f"Built {index_path}")
    print(f"Zipped to {zip_path}")

    # If running in Google Colab, trigger a download automatically
    try:
        from google.colab import files  # type: ignore
        files.download(zip_path)
    except ImportError:
        print("(Not running in Colab — find your files in the 'portfolio_site' folder "
              "and 'portfolio_site.zip' next to this script.)")


if __name__ == "__main__":
    main()
```
